```python
# 환경 변수 불러오기
from dotenv import load_dotenv

load_dotenv()
```

```python
# 클라이언트 생성
from anthropic import Anthropic

# dotenv로 불러오면 api key 명시하지 않아도 되는 것 같음
client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# 유틸 함수 만들기

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )

    return message.content[0].text
```

```python
# 메세지 리스트
messages = []

# 초기 유저의 질문을 추가한다.
add_user_message(messages, "양자 컴퓨팅에 대해 한 문장으로 알려줘.")

# 메세지 리스트를 chat에 넣어 답변을 구한다.
answer = chat(messages)

# 받은 답변도 메세지 리스트에 추가한다.
add_assistant_message(messages, answer)

# 유저의 후속 질문을 메세지 리스트에 추가한다.
add_user_message(messages, "한 문장 더 작성해서 추가 설명해봐.")

# 다시 chat을 호출하여 최종 답변을 얻는다.
answer = chat(messages)

print("메세지 리스트: ", messages)
print("최종 답변: ", answer)
```
