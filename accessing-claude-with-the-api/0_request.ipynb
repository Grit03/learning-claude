```python
# 환경 변수 불러오기
from dotenv import load_dotenv

load_dotenv()
```

```python
# 클라이언트 생성
from anthropic import Anthropic

# dotenv로 불러오면 api key 명시하지 않아도 되는 것 같음
client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# 요청 보내기
# anthropic sdk의 create 함수를 통해 claude에 접근할 수 있다.
# 이 함수는 다음 항목들을 필요로 한다.
# model: 사용하려는 Claude 모델 이름
# max_tokens: 응답 최대 길이(안전 한계) (claude가 생성할 수 있는 토큰 수에 대한 최대 예산)
# messages: Claude에게 보낼 대화 내역

message = client.messages.create(
    model=model,
    max_tokens=1000,
    # 입력 메세지
    messages=[
        {"role": "user", "content": "양자 컴퓨팅이 무엇인지, 한 문장으로 말해줘."}
    ]
)
```
